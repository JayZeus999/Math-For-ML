```python
import matplotlib.colors as mcolors
import matplotlib.pyplot as plt
```

```python
print(mcolors.CSS4_COLORS.keys())
```

## PLOTLINE GALLERY

```python
colors = mcolors.CSS4_COLORS
plt.figure(figsize=(10, 40))

for i, (name, color) in enumerate(colors.items()):
    plt.barh(i, 1, color=color)
    plt.text(0.02, i, name, va="center")

plt.yticks([])
plt.xticks([])
plt.show()
```

## COLORMAP GALLERY

```python
import matplotlib.pyplot as plt
import matplotlib as mpl

cmaps = plt.colormaps()

fig, ax = plt.subplots(figsize=(8, len(cmaps) * 0.25))

for i, cmap in enumerate(cmaps):
    gradient = [[0, 1]]
    ax.imshow(
        gradient,
        aspect="auto",
        cmap=mpl.colormaps[cmap],
        extent=[0, 1, i, i + 0.8]
    )
    ax.text(1.02, i + 0.4, cmap, va="center")

ax.set_xlim(0, 1)
ax.set_ylim(0, len(cmaps))
ax.axis("off")

plt.show()
```
